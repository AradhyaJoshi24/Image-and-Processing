```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.png", cv2.IMREAD_GRAYSCALE)

threshold = int(input("Enter threshold value (0-255): "))

_, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")

plt.show()

cv2.imwrite("binary_user.png", binary)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.png")

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
gray = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale (Average)")
plt.axis("off")

plt.show()

cv2.imwrite("gray_average.png", gray)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image.png")

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

wr = float(input("Enter weight for Red: "))
wg = float(input("Enter weight for Green: "))
wb = float(input("Enter weight for Blue: "))

if wr < 0 or wg < 0 or wb < 0 or wr > 1 or wg > 1 or wb > 1:
    print("Error: Weightages must be between 0 and 1.")

elif abs((wr + wg + wb) - 1) > 1e-6:
    print("Error: Sum of weightages must be equal to 1.")

else:
    R = img[:, :, 0].astype(float)
    G = img[:, :, 1].astype(float)
    B = img[:, :, 2].astype(float)

    gray = (wr * R + wg * G + wb * B).astype(np.uint8)

    plt.figure(figsize=(8,4))

    plt.subplot(1,2,1)
    plt.imshow(img)
    plt.title("Original RGB")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(gray, cmap='gray')
    plt.title("Weighted Grayscale")
    plt.axis("off")
    
    plt.show()
    
    cv2.imwrite("gray_weighted.png", gray)
```
