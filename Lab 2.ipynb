```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("image.png", cv2.IMREAD_GRAYSCALE)

# Check image
if img is None:
    print("Error: image.png not found!")
else:
    # Calculate mean intensity
    threshold = np.mean(img)

    print("Mean Threshold =", threshold)

    # Convert grayscale to binary
    _, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)

    # Display images
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(binary, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    plt.show()

    # Save binary image
    cv2.imwrite("binary_mean.png", binary)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.png")

# Check image
if img is None:
    print("Error: image.png not found!")
else:
    # Convert BGR to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Separate R, G and B planes
    R = rgb[:, :, 0].astype(float)
    G = rgb[:, :, 1].astype(float)
    B = rgb[:, :, 2].astype(float)

    # Mean average of R, G and B
    gray = ((R + G + B) / 3).astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(rgb)
    plt.title("Original RGB Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.show()

    # Save grayscale image
    cv2.imwrite("gray_average.png", gray)
```

```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.png")

# Check image
if img is None:
    print("Error: image.png not found!")
else:
    # Convert BGR to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Input weightages
    R_weight = float(input("Enter weightage of R: "))
    G_weight = float(input("Enter weightage of G: "))
    B_weight = float(input("Enter weightage of B: "))

    # Validate weightages
    if (R_weight < 0 or R_weight > 1 or
        G_weight < 0 or G_weight > 1 or
        B_weight < 0 or B_weight > 1):

        print("Error: Each weightage must be between 0 and 1.")

    elif abs(R_weight + G_weight + B_weight - 1) > 0.000001:

        print("Error: Sum of weightages must be equal to 1.")

    else:
        # Separate R, G and B planes
        R = rgb[:, :, 0]
        G = rgb[:, :, 1]
        B = rgb[:, :, 2]

        # Calculate weighted grayscale
        gray = (
            R_weight * R +
            G_weight * G +
            B_weight * B
        ).astype("uint8")

        # Display images
        plt.figure(figsize=(10, 4))

        plt.subplot(1, 2, 1)
        plt.imshow(rgb)
        plt.title("Original RGB Image")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(gray, cmap="gray")
        plt.title("Weighted Grayscale Image")
        plt.axis("off")

        plt.show()

        # Save grayscale image
        cv2.imwrite("gray_weighted.png", gray)
```
